# Statistical Analysis
Five requested tests: paired t-test, one-sample t-test, Pearson, Spearman, and delay z-scores.

```python
from scipy import stats
import numpy as np, pandas as pd
from src.feature_engineering import build_features
df=build_features(); df['shipment_dt']=pd.to_datetime(df.shipment_dt); df['actual_dt']=pd.to_datetime(df.actual_dt); df['delivery_hours']=(df.actual_dt-df.shipment_dt).dt.total_seconds()/3600
q=df.dropna(subset=['delivery_hours']).copy(); q['quarter']=q.shipment_dt.dt.quarter; q=q[q.quarter.isin([1,2])]; p=q.groupby(['driver_id','quarter']).delivery_hours.mean().unstack().dropna()
print('Paired t-test Q1 vs Q2:',stats.ttest_rel(p[1],p[2]) if len(p)>1 else 'Insufficient paired observations')
r=df.rating.dropna(); print('One sample rating vs 4:',stats.ttest_1samp(r,4.0) if len(r)>1 else 'Insufficient data')
a=df[['distance_km','delay_hours']].dropna(); print('Pearson:',stats.pearsonr(a.distance_km,a.delay_hours) if len(a)>2 else 'Insufficient data')
a=df[['experience','rating']].dropna(); print('Spearman:',stats.spearmanr(a.experience,a.rating) if len(a)>2 else 'Insufficient data')
x=df.delay_hours.dropna(); z=(x-x.mean())/x.std(ddof=0); print('Z-score outliers |z|>3:',int((z.abs()>3).sum()))
```
